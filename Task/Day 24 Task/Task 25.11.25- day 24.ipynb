```python
import pandas as pd
from sklearn.impute import SimpleImputer
```

```python
df = pd.DataFrame({
    'A': [1, 2, None, 4],
    'B': [None, 5, 6, None],
    'C': [7, None, 9, 10]
})

print("Original DataFrame:")
print(df)
```

```python
print(df.isnull())
```

```python
print(df.isnull().sum())
```

```python
print(df.isnull().sum().sum())
```

```python
df_dropna = df.dropna()
print("\n4. df.dropna() (Rows with any NaN removed)")
print(df_dropna)
```

```python
df_dropna_all = df.dropna(how = 'all')
print(df_dropna_all)
```

```python
df_drop_col = df.dropna(axis=1)
print("\n6. df.dropna(axis=1) (Drop columns with NaN)")
print(df_drop_col)
```

```python
df_fill0 = df.fillna(0)
print("\n7. df.fillna(0)")
print(df_fill0)
```

```python
df_mean = df.copy()
df_mean['A'].fillna(df_mean['A'].mean(), inplace=True)
print("\n8. Fill NaN in column A with mean:")
print(df_mean)
```

```python
df_mode = df.copy()
df_mode['B'].fillna(df_mode['B'].mode()[0], inplace=True)
print("\n9. Fill NaN in column B with mode:")
print(df_mode)
```

```python
df_interpolate = df.copy()
df_interpolate.interpolate(inplace=True)
print("\n10. df.interpolate()")
print(df_interpolate)
```

```python
imputer = SimpleImputer(strategy='median')

df_imputed = pd.DataFrame(imputer.fit_transform(df), columns=df.columns)
print("\n11 & 12. SimpleImputer(strategy='median')")
print(df_imputed)
```
