```python
# 1. Import libraries
import numpy as np
import pandas as pd
from scipy.stats import norm
```

```python
population = np.random.rand(10000)
```

```python
sample = np.random.choice(population, size=100)
```

```python
df = pd.DataFrame(population, columns=['value'])
```

```python
df['category'] = np.random.choice(['A', 'B', 'C'], size=len(df))
```

```python
sample_n = df.sample(n=50)
sample_frac = df.sample(frac=0.1)
```

```python
sample_means = [np.mean(np.random.choice(population, 30)) for _ in range(1000)]
```

```python
# 7. Groupby sampling
group_samples = df.groupby('category').apply(lambda x: x.sample(n=10))
```

```python
# 8. Calculate z-score
z_score = (sample.mean() - population.mean()) / (population.std() / np.sqrt(len(sample)))
```

```python
# 9. Calculate p-value
p_value = norm.cdf(z_score)
```

```python
 #10. Calculate Standard Error of sample means
standard_error = np.std(sample_means)
```

```python
# Print outputs
print("Sample (n=100):", sample[:5])
print("Random sample from df (n=50):")
print(sample_n.head())

print("\nRandom sample (10%):")
print(sample_frac.head())

print("\nGroupby samples:")
print(group_samples.head())

print("\nZ-score:", z_score)
print("P-value:", p_value)
print("Standard Error of Sampling Distribution:", standard_error)
```
