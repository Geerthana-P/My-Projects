```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Using exponential distribution (highly skewed)
population_size = 100000
population = np.random.exponential(scale=2, size=population_size)
```

```python
# 2. Simulation Loop for CLT
# ---------------------------------------------
sample_means = []
iterations = 1000        # Number of times we take samples
sample_size = 30         # Each sample size

for _ in range(iterations):
    sample = np.random.choice(population, size=sample_size)
    mean_value = np.mean(sample)
    sample_means.append(mean_value)

sample_means = np.array(sample_means)
```

```python
# 3. Visualization
# ---------------------------------------------

plt.figure(figsize=(14,6))
```

```python
# Histogram of original (non-normal) population
plt.subplot(1, 2, 1)
plt.hist(population, bins=50, edgecolor='black')
plt.title("Original Population (Skewed Distribution)")
plt.xlabel("Value")
plt.ylabel("Frequency")
```

```python
# Histogram of sample means (CLT effect)
plt.subplot(1, 2, 2)
plt.hist(sample_means, bins=30, edgecolor='black')
plt.title("Distribution of Sample Means (CLT Bell Curve)")
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```
