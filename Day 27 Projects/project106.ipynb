```python
import numpy as np
from scipy import stats
```

```python
# Set seed for reproducibility
np.random.seed(42)

# Generate survey ratings (slightly higher than neutral)
ratings = np.random.normal(loc=3.4, scale=0.6, size=40)

# Optional: clip values to keep them within 1–5 scale
ratings = np.clip(ratings, 1, 5)

# View first 10 ratings
ratings[:10]
```

```python
print("Sample Mean Rating:", np.mean(ratings))
print("Sample Standard Deviation:", np.std(ratings, ddof=1))
print("Number of Responses:", len(ratings))
```

```python
t_statistic, p_value = stats.ttest_1samp(ratings, popmean=3)

print("T-statistic:", t_statistic)
print("P-value:", p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject H₀: The average survey rating is significantly different from 3.")
else:
    print("Fail to Reject H₀: No significant difference from the neutral rating.")
```
