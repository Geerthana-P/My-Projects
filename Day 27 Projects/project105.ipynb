```python
import numpy as np
from scipy import stats
```

```python
# Set seed for reproducibility
np.random.seed(42)

# Generate performance scores
before_training = np.random.normal(loc=65, scale=8, size=20)
after_training = before_training + np.random.normal(loc=7, scale=3, size=20)

# View first 10 scores
before_training[:10], after_training[:10]
```

```python
print("Before Training Mean:", np.mean(before_training))
print("After Training Mean:", np.mean(after_training))
print("Mean Improvement:", np.mean(after_training - before_training))
```

```python
t_statistic, p_value = stats.ttest_rel(after_training, before_training)

print("T-statistic:", t_statistic)
print("P-value:", p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject H₀: Training significantly improved employee performance.")
else:
    print("Fail to Reject H₀: No significant improvement detected.")
```
