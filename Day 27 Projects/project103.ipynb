```python
import numpy as np
from scipy import stats
```

```python
# Set seed for reproducibility
np.random.seed(42)

# Generate sample data
sample_weights = np.random.normal(loc=185, scale=10, size=50)

# View first 10 values
sample_weights[:10]
```

```python
print("Sample Mean:", np.mean(sample_weights))
print("Sample Standard Deviation:", np.std(sample_weights, ddof=1))
print("Sample Size:", len(sample_weights))
```

```python
t_statistic, p_value = stats.ttest_1samp(sample_weights, popmean=180)

print("T-statistic:", t_statistic)
print("P-value:", p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject H₀: The mean weight is statistically different from 180 lbs.")
else:
    print("Fail to Reject H₀: No significant difference from 180 lbs.")
```
