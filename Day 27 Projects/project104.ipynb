```python
import numpy as np
from scipy import stats
```

```python
# Set seed for reproducibility
np.random.seed(42)

# Generate scores
tutoring_scores = np.random.normal(loc=78, scale=8, size=40)
non_tutoring_scores = np.random.normal(loc=72, scale=8, size=40)

# View first 10 scores from each group
tutoring_scores[:10], non_tutoring_scores[:10]
```

```python
print("Tutoring Group Mean:", np.mean(tutoring_scores))
print("Non-Tutoring Group Mean:", np.mean(non_tutoring_scores))
print("Tutoring Group Std Dev:", np.std(tutoring_scores, ddof=1))
print("Non-Tutoring Group Std Dev:", np.std(non_tutoring_scores, ddof=1))
```

```python
t_statistic, p_value = stats.ttest_ind(
    tutoring_scores,
    non_tutoring_scores,
    equal_var=False  # Welch's t-test
)

print("T-statistic:", t_statistic)
print("P-value:", p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject H₀: Tutoring has a statistically significant positive effect on scores.")
else:
    print("Fail to Reject H₀: No significant difference between the groups.")
```
