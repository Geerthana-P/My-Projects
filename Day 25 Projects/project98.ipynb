```python
import pandas as pd
```

```python
data = {
    'customer_id': [101, 101, 102, 103, 103, 104],
    'name': ['John Doe', ' john doe ', 'ALICE SMITH', 'Alice smith', 'Alice  Smith', 'Bob Ray'],
    'address': ['New York', ' New york ', 'London', 'LONDON ', ' london', 'Delhi']
}

df = pd.DataFrame(data)

print("Original Customer Data:")
print(df)
print("\n")
```

```python
df = df.drop_duplicates(subset='customer_id', keep='first')

print("After Removing Exact Duplicates Based on customer_id:")
print(df)
print("\n")
```

```python
df['name'] = df['name'].str.strip()
df['address'] = df['address'].str.strip()
```

```python
df['name'] = df['name'].str.upper()
df['address'] = df['address'].str.upper()

print("After Standardizing Name and Address:")
print(df)
print("\n")
```

```python
print("Occurrences of Each Standardized Name:")
print(df.groupby('name').size().sort_values(ascending=False))
print("\n")

# If any fuzzy duplicates exist (same name & address) → remove
df = df.drop_duplicates(subset=['name', 'address'], keep='first')
```

```python
print("Final Cleaned Customer Database:")
print(df)
print("\n")

# Re-check for duplicates
print("Remaining Duplicates:", df.duplicated().sum())
```
