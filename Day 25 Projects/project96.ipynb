```python
import pandas as pd
```

```python
data = {
    'name': ['John Doe', 'john doe', 'Alice Smith', 'Alice Smith', 'Bob Ray', 'bob ray'],
    'age': [25, 25, 30, 30, 22, 22],
    'city': ['New York', 'New York', 'London', 'London', 'Delhi', 'Delhi']
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)
print("\n")
```

```python
print("Duplicate Row Flags:")
print(df.duplicated())

print("\nNumber of Exact Duplicates:", df.duplicated().sum())
print("\n")
```

```python
df.drop_duplicates(inplace=True)

print("DataFrame After Removing Exact Duplicates:")
print(df)
print("\n")
```

```python
# Convert names to lowercase
df['name'] = df['name'].str.lower()

print("DataFrame After Name Standardization:")
print(df)
print("\n")

# Check counts of standardized names
print("Counts of Standardized Names:")
print(df.groupby('name').size().sort_values(ascending=False))
print("\n")
```

```python
df.drop_duplicates(inplace=True)
```

```python
print("Final Number of Duplicates:", df.duplicated().sum())
print("\nFinal Cleaned DataFrame:")
print(df)
```
