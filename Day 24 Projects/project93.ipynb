```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# 1. Create a Sample Time-Series Dataset
# --------------------------------------------
data = {
    "date": pd.date_range(start="2024-01-01", periods=10, freq="D"),
    "temperature": [30, 31, None, 29, None, 28, 30, None, 32, 33]
}

df = pd.DataFrame(data)
df.set_index('date', inplace=True)

print("Original Time-Series Data (with missing values):")
print(df)
```

```python
print("\nMissing values:")
print(df.isnull().sum())
```

```python
df_interpolated = df.interpolate(method='time')

print("\nAfter Time-Based Interpolation:")
print(df_interpolated)
```

```python
plt.figure(figsize=(10,5))
plt.plot(df.index, df['temperature'], marker='o')
plt.title("Before Imputation")
plt.show()

plt.figure(figsize=(10,5))
plt.plot(df_interpolated.index, df_interpolated['temperature'], marker='o')
plt.title("After Time-Based Imputation")
plt.show()
```
