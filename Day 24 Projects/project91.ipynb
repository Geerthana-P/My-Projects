```python
import pandas as pd

df = pd.read_csv("sample.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
df['age'].hist()
```

```python
df['age'].fillna(df['age'].mean(), inplace=True)
```

```python
df['salary'].fillna(df['salary'].median(), inplace=True)
```

```python
df['city'].value_counts()
```

```python
df['city'].fillna(df['city'].mode()[0], inplace=True)
```
