```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv("students.csv")
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x="Gender", y="Math")
plt.title("Math Scores by Gender")
plt.xlabel("Gender")
plt.ylabel("Math Score")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.violinplot(data=df, x="StudyGroup", y="Science")
plt.title("Science Score Distribution by Study Group")
plt.xlabel("Study Group")
plt.ylabel("Science Score")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Math", y="Science", hue="Gender")
plt.title("Relationship Between Math and Science Scores")
plt.xlabel("Math Score")
plt.ylabel("Science Score")
plt.show()
```
