```python
import pandas as pd
from scipy.stats import chi2_contingency
```

```python
data = {
    "region": (
        ["North"] * 25 +
        ["South"] * 25 +
        ["East"] * 25 +
        ["West"] * 25
    ),
    "rating": (
        # North
        ["5-star"] * 10 + ["4-star"] * 8 + ["3-star"] * 5 + ["2-star"] * 2 +
        # South
        ["3-star"] * 8 + ["4-star"] * 7 + ["2-star"] * 6 + ["1-star"] * 4 +
        # East
        ["5-star"] * 12 + ["4-star"] * 8 + ["3-star"] * 5 +
        # West
        ["4-star"] * 10 + ["3-star"] * 8 + ["2-star"] * 5 + ["1-star"] * 2
    )
}

df = pd.DataFrame(data)
df.head()
```

```python
df["region"].value_counts(), df["rating"].value_counts()
```

```python
contingency_table = pd.crosstab(df["region"], df["rating"])
contingency_table
```

```python
chi2_stat, p_value, dof, expected = chi2_contingency(contingency_table)

print("Chi-Square Statistic:", chi2_stat)
print("P-value:", p_value)
print("Degrees of Freedom:", dof)
```

```python
expected_df = pd.DataFrame(
    expected,
    index=contingency_table.index,
    columns=contingency_table.columns
)

expected_df
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
    print("There IS a statistically significant association between customer rating and region.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("There is NO statistically significant association between customer rating and region.")
```
