```python
import pandas as pd
from scipy.stats import chi2_contingency
```

```python
# Creating sample data with a slight association
data = {
    "education_level": (
        ["High School"] * 40 +
        ["College"] * 40 +
        ["Graduate"] * 40
    ),
    "homeowner": (
        ["Yes"] * 10 + ["No"] * 30 +     # High School
        ["Yes"] * 20 + ["No"] * 20 +     # College
        ["Yes"] * 30 + ["No"] * 10       # Graduate
    )
}

df = pd.DataFrame(data)
df.head()
```

```python
df["education_level"].value_counts(), df["homeowner"].value_counts()
```

```python
contingency_table = pd.crosstab(df["education_level"], df["homeowner"])
contingency_table
```

```python
chi2_stat, p_value, dof, expected = chi2_contingency(contingency_table)

print("Chi-Square Statistic:", chi2_stat)
print("P-value:", p_value)
print("Degrees of Freedom:", dof)
```

```python
expected_df = pd.DataFrame(
    expected,
    index=contingency_table.index,
    columns=contingency_table.columns
)

expected_df
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
    print("There IS a statistically significant relationship between education level and homeownership.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("There is NO statistically significant relationship between education level and homeownership.")
```
