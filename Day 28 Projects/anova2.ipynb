```python
import numpy as np
from scipy.stats import f_oneway
```

```python
# Durability data in hours
supplier_A = np.array([520, 530, 515, 525, 528])
supplier_B = np.array([540, 545, 538, 542, 547])  # Higher mean
supplier_C = np.array([518, 522, 520, 519, 521])
supplier_D = np.array([525, 527, 523, 526, 524])

print("Supplier A Mean:", supplier_A.mean())
print("Supplier B Mean:", supplier_B.mean())
print("Supplier C Mean:", supplier_C.mean())
print("Supplier D Mean:", supplier_D.mean())
```

```python
print("Null Hypothesis (H0): Average durability is the same for all four suppliers.")
print("Alternative Hypothesis (Ha): At least one supplier has a different average durability.")
```

```python
f_statistic, p_value = f_oneway(
    supplier_A,
    supplier_B,
    supplier_C,
    supplier_D
)

print("F-statistic:", f_statistic)
print("P-value:", p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
    print("There IS a statistically significant difference in average durability among suppliers.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("There is NO statistically significant difference in average durability among suppliers.")
```
