```python
import numpy as np
from scipy.stats import f_oneway
```

```python
# Weight loss data in kilograms
diet_A = np.array([2.1, 2.5, 2.3, 2.0, 2.4])
diet_B = np.array([3.5, 3.8, 3.6, 3.7, 3.9])  # Higher mean
diet_C = np.array([2.2, 2.4, 2.1, 2.3, 2.5])

print("Diet A Mean:", diet_A.mean())
print("Diet B Mean:", diet_B.mean())
print("Diet C Mean:", diet_C.mean())
```

```python
print("Null Hypothesis (H0): Mean weight loss is the same for all three diets.")
print("Alternative Hypothesis (Ha): At least one diet has a different mean weight loss.")
```

```python
f_statistic, p_value = f_oneway(diet_A, diet_B, diet_C)

print("F-statistic:", f_statistic)
print("P-value:", p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
    print("There IS a statistically significant difference in average weight loss among the diets.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("There is NO statistically significant difference in average weight loss among the diets.")
```
