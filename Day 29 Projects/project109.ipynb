```python
import pandas as pd
from itertools import combinations

```

```python
# Load dataset
df = pd.read_csv("sales_transactions.csv")

# View first few rows
df.head()
```

```python
df.info()
```

```python
df['order_id'].nunique()
```

```python
# Group products purchased in the same order
grouped = df.groupby('order_id')['product'].apply(list)

grouped.head()
```

```python
pair_list = []

for products in grouped:
    # Create all unique product pairs
    pairs = combinations(sorted(products), 2)
    pair_list.extend(pairs)
```

```python
pair_counts = pd.Series(pair_list).value_counts()

pair_counts.head(10)
```

```python
pair_df = pair_counts.reset_index()
pair_df.columns = ['Product Pair', 'Frequency']

pair_df.head()
```
