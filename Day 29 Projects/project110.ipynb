```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Study hours (independent variable)
study_hours = np.array([2, 4, 6, 8, 10])

# Final scores (dependent variable)
final_score = np.array([50, 55, 65, 70, 80])
```

```python
mean_x = np.mean(study_hours)
mean_y = np.mean(final_score)

print("Mean Study Hours:", mean_x)
print("Mean Final Score:", mean_y)
```

```python
numerator = np.sum((study_hours - mean_x) * (final_score - mean_y))
denominator = np.sqrt(
    np.sum((study_hours - mean_x) ** 2) *
    np.sum((final_score - mean_y) ** 2)
)

correlation = numerator / denominator
print("Correlation Coefficient:", correlation)
```

```python
plt.scatter(study_hours, final_score)
plt.xlabel("Study Hours")
plt.ylabel("Final Score")
plt.title("Study Hours vs Final Score")
plt.show()
```

```python
m = 0   # slope
b = 0   # intercept
learning_rate = 0.01
epochs = 1000
n = len(study_hours)
```

```python
for i in range(epochs):
    y_pred = m * study_hours + b
    
    dm = (2/n) * np.sum((y_pred - final_score) * study_hours)
    db = (2/n) * np.sum(y_pred - final_score)
    
    m = m - learning_rate * dm
    b = b - learning_rate * db

print("Slope (m):", m)
print("Intercept (b):", b)
```

```python
y_predicted = m * study_hours + b

plt.scatter(study_hours, final_score)
plt.plot(study_hours, y_predicted)
plt.xlabel("Study Hours")
plt.ylabel("Final Score")
plt.title("Linear Regression using Gradient Descent")
plt.show()
```

```python
print(f"Final Score = {m:.2f} * Study Hours + {b:.2f}")
```
