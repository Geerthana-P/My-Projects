```python
import pandas as pd
import numpy as np
from scipy.stats import ttest_ind
```

```python
# Load sales dataset
df = pd.read_csv("sales_campaign_data.csv")

# Display first few rows
df.head()
```

```python
df.info()
```

```python
df['campaign_group'].value_counts()
```

```python
print("Problem Statement:")
print("Does the new marketing campaign generate more revenue?\n")

print("Null Hypothesis (H0):")
print("Average revenue of Test group = Average revenue of Control group\n")

print("Alternative Hypothesis (Ha):")
print("Average revenue of Test group > Average revenue of Control group")
```

```python
# Separate revenue data for Test and Control groups
test_group_rev = df[df['campaign_group'] == 'Test']['revenue']
control_group_rev = df[df['campaign_group'] == 'Control']['revenue']

print("Test Group Sample Size:", test_group_rev.shape[0])
print("Control Group Sample Size:", control_group_rev.shape[0])
```

```python
print("Test Group Mean Revenue:", test_group_rev.mean())
print("Control Group Mean Revenue:", control_group_rev.mean())
```

```python
t_stat, p_value = ttest_ind(
    test_group_rev,
    control_group_rev,
    alternative='greater'
)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis (H0)")
    print("Conclusion: The marketing campaign had a statistically significant positive impact on revenue.")
else:
    print("Fail to Reject the Null Hypothesis (H0)")
    print("Conclusion: There is no statistically significant evidence that the campaign increased revenue.")
```
