```python
import pandas as pd
```

```python
df = pd.read_csv("sales.csv")

print("Dataset Loaded Successfully!\n")
print("First 5 Rows:")
display(df.head())
```

```python
if "revenue" not in df.columns:
    df["revenue"] = df["units_sold"] * df["price"]

print("\nRevenue column added (if it was missing)")
```

```python
top_products = (
    df.groupby("product")["units_sold"]
      .sum()
      .sort_values(ascending=False)
)

print("\nTop-Selling Products:")
display(top_products)
```

```python
avg_revenue_region = (
    df.groupby("region")["revenue"]
      .mean()
      .sort_values(ascending=False)
)

print("\nAverage Revenue Per Region:")
display(avg_revenue_region)
```

```python
top_products.to_csv("top_selling_products.csv")
avg_revenue_region.to_csv("average_revenue_per_region.csv")

print("\nAnalysis Completed Successfully!")
```
