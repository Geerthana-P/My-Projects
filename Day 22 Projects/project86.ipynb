```python
import pandas as pd
```

```python
df = pd.read_csv("movies.csv")

print("Original Dataset:")
display(df.head())
```

```python
print("\nNumber of rows before dropping duplicates:", len(df))
df = df.drop_duplicates()
print("Number of rows after dropping duplicates:", len(df))
```

```python
df["genre"] = df["genre"].str.lower()
```

```python
df["genre"] = df["genre"].str.strip()
```

```python
df["genre"] = df["genre"].str.replace("science fiction", "sci-fi")
df["genre"] = df["genre"].str.replace("sci fi", "sci-fi")
df["genre"] = df["genre"].str.replace("sci fi", "sci-fi")
df["genre"] = df["genre"].str.replace("action/adventure", "action")
df["genre"] = df["genre"].str.replace("romantic", "romance")
```

```python
df["genre"] = df["genre"].str.replace(r"\s+", " ", regex=True)

print("\nCleaned Genre Column:")
display(df["genre"].unique())
```

```python
df.to_csv("movies_cleaned.csv", index=False)
print("\nMovie dataset cleaned and saved as movies_cleaned.csv")
```
