```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Display settings
%matplotlib inline
plt.rcParams['figure.figsize'] = (10, 6)
sns.set(style="whitegrid")
```

```python
df = sns.load_dataset("titanic")
print("Dataset Loaded Successfully!\n")
display(df.head())
```

```python
print("Missing Value Count:\n")
print(df.isnull().sum())
```

```python
plt.figure(figsize=(12, 6))
sns.heatmap(df.isnull(), cbar=False, yticklabels=False)
plt.title("Missing Values Heatmap")
plt.show()
```

```python
df['age'].fillna(df['age'].median(), inplace=True)
```

```python
df.drop(columns=['deck'], inplace=True)
```

```python
df['embarked'].fillna(df['embarked'].mode()[0], inplace=True)
```

```python
df['embark_town'].fillna(df['embark_town'].mode()[0], inplace=True)
```

```python
df['family_size'] = df['sibsp'] + df['parch']
```

```python
df['is_alone'] = df['family_size'].apply(lambda x: 1 if x == 0 else 0)
```

```python
print("\nAfter Cleaning Missing Values:\n")
print(df.isnull().sum())
```

```python
plt.figure(figsize=(10,5))
sns.histplot(df['age'], kde=True)
plt.title("Age Distribution After Median Imputation")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='embarked', data=df)
plt.title("Embarked Distribution (After Cleaning)")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='is_alone', data=df)
plt.title("Is_Alone Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x='is_alone', y='survived', data=df)
plt.title("Survival Rate: Alone vs With Family")
plt.show()

print("\nEDA & Cleaning Completed Successfully!")
```
