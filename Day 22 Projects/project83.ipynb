```python
import pandas as pd, numpy as np
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
plt.rcParams['figure.figsize'] = (8,5)
sns.set(style="whitegrid")
```

```python
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = pd.Categorical.from_codes(iris.target, iris.target_names)
```

```python
display(df.head())
display(df.tail())
print(df.info())
print("\nMissing values:\n", df.isna().sum())
```

```python
display(df.describe().T)
print("\nSpecies counts:\n", df['species'].value_counts())
```

```python
sns.countplot(x='species', data=df); plt.title('Species counts'); plt.show()
df.iloc[:, :4].hist(bins=12); plt.suptitle('Feature histograms'); plt.show()
sns.boxplot(data=df.iloc[:, :4]); plt.title('Boxplots'); plt.show()
sns.pairplot(df, hue='species', diag_kind='hist'); plt.suptitle('Pairplot', y=1.02); plt.show()
```

```python
corr = df.iloc[:, :4].corr()
display(corr)
sns.heatmap(corr, annot=True, fmt=".2f", cmap='vlag', square=True); plt.title('Correlation matrix'); plt.show()

display(df.groupby('species').mean())
```
